Create a Simple Line Plot

Dataset:

x = [1,2,3,4,5]
y = [10,20,30,40,50]

Tasks:

Plot line graph
Add title
Add x-label and y-label

Concepts:

plt.plot()
plt.title()
plt.xlabel()
plt.ylabel()

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.plot(x,y)
plt.title("X vs Y")
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
```

Plot Multiple Lines

Tasks:

Plot sales of Product A
Plot sales of Product B
Add legend

Concepts:

Multiple plots
plt.legend()

```python
months = ['Jan','Feb','Mar','Apr','May']
sales_A = [150, 200, 180, 220, 300]
sales_B = [180, 190, 210, 240, 260]
plt.plot(months, sales_A, label='Product A', marker='o', linewidth=2)
plt.plot(months, sales_B, label='Product B', marker='s', linewidth=2)
plt.xlabel('Months')
plt.ylabel('Sales(unit)')
plt.title('Month Sales Comparison')
plt.legend()
plt.show()
```

Create a Histogram

Tasks:

Generate random marks
Plot distribution

Concepts:

plt.hist()

```python
total=['year1','year2','year3','year4','year5']
marks=[78,87,89,86,85]
plt.hist(total, bins=len(total), weights=marks, color='blue', edgecolor='black')
plt.xlabel('total')
plt.ylabel('marks')
plt.title('Marks over the Year')
plt.show()
```

Create a Pie Chart

Tasks:

Show budget distribution

Example:

Food = 40
Rent = 30
Travel = 20
Other = 10

Concepts:

plt.pie()

```python
categories=['food','rent','travel','other']
budget=[40,30,45,55]
color = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
plt.pie(budget, labels=categories, autopct='%1.1f%%', startangle=140, colors=color)
plt.axis('equal')
plt.title('Budget Distribution')
plt.show()
```

```python
x=[1,2,3,4,5]
y=[2,4,1,5,3]
plt.plot(x,y, marker='D', linestyle='--',color='purple', markersize=8)
plt.xlabel('X Axis')
plt.ylabel('Y label')
plt.title('Customized Line Plot')
plt.grid(True, color='gray', linestyle=':',linewidth=0.5)
plt.show()
```

Student Performance Dashboard

Dataset:

Student	Marks
A	75
B	82
C	95
D	67

Tasks:

Bar chart
Average line
Highest scorer highlight

Concepts:

Multiple plotting elements

```python
students = ['A', 'B', 'C', 'D']
marks = [75, 82, 95, 67]

avg_mark = sum(marks) / len(marks)
max_mark = max(marks)
max_student = students[marks.index(max_mark)]

plt.figure(figsize=(8, 5))

colors = ['#1f77b4' if mark != max_mark else '#2ca02c' for mark in marks]
bars = plt.bar(students, marks, color=colors)

plt.axhline(y=avg_mark, color='red', linestyle='--', linewidth=2, label=f'Average ({avg_mark:.1f})')

plt.annotate(f'Highest: {max_mark}', 
             xy=(max_student, max_mark), 
             xytext=(max_student, max_mark - 15),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6),
             ha='center', color='green', weight='bold')

plt.xlabel('Students', fontsize=12)
plt.ylabel('Marks', fontsize=12)
plt.title('Student Performance Dashboard', fontsize=14, weight='bold')
plt.ylim(0, 110)  
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.legend()
plt.show()
```
