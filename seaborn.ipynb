Create a Seaborn Line Plot

Dataset:

days = [1,2,3,4,5]
sales = [100,120,140,130,170]

Concepts:

sns.lineplot()

```python

import seaborn as sns
import matplotlib.pyplot as plt
days = [1,2,3,4,5]
sales = [100,120,140,130,170]
sns.lineplot( x=days, y=sales, marker='o')
plt.xlabel('Days')
plt.ylabel('Sales')
plt.title('Daily sales Tracking')

```

Create a Count Plot

Dataset:

Male/Female students

Tasks:

Count categories

Concepts:

sns.countplot()

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
data = {
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female', 'Male', 'Female', 'Female', 'Male', 'Female', 'Male', 'Female']
}
df = pd.DataFrame(data)
category_order = df['Gender'].value_counts().index
sns.countplot(data=df, x='Gender', order=category_order)
plt.xlabel('Gender')
plt.ylabel('Student Count')
plt.title('Distribution of Male and Female Students')
plt.savefig('student_count_plot.png', bbox_inches='tight')
```

Create Histogram

Tasks:

Show marks distribution

Concepts:

sns.histplot()

```python
import numpy as np
np.random.seed(42)
marks = np.random.normal(loc=75, scale=12, size=150)
marks = np.clip(marks, 0, 100)
sns.histplot(x=marks, bins=15, kde=True, color='skyblue')
plt.xlabel('Marks')
plt.ylabel('Number of Students')
plt.title('Distribution of Students Marks')
plt.savefig('marks_histogram.png', bbox_inches='tight')
```

Create Box Plot

Tasks:

Detect outliers

Concepts:

sns.boxplot()

```python
import pandas as pd
np.random.seed(10)
normal_marks = np.random.normal(loc=72, scale=8, size=100)
outliers = np.array([20, 25, 30, 99, 100]) 
all_marks = np.concatenate([normal_marks, outliers])
df = pd.DataFrame({'Marks': all_marks})
sns.boxplot(x=df['Marks'], color='lightgreen')
plt.xlabel('Student Marks')
plt.title('Outlier Detection in Student Marks')
plt.show()
```

Scatter Plot with Categories

Tasks:

Marks vs Study Hours
Color by Gender

Concepts:

sns.scatterplot()

```python
np.random.seed(42)
n_samples =60
study_hour = np.random.uniform(1,10,n_samples)
marks = np.clip(35 + (study_hour*5.5) + np.random.normal(0,6,n_samples),0,100)
gender = np.random.choice(['Male','Female'], size=n_samples)
df = pd.DataFrame({
    'Study Hours': study_hour,
    'Marks': marks,
    'Gender': gender
})
sns.scatterplot(data=df, x='Study Hours', y='Marks', hue='Gender', palette='Set1', s=70)
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Marks vs. Study Hours (Grouped by Gender)')
```

Pair Plot

```python
data = pd.DataFrame({
    "Age": [18,19,20,21,22],
    "Marks": [80,90,70,85,95],
    "Hours_Studied": [5,6,4,5,7]
})

sns.pairplot(data)
```
