GroupBy Operations

Perform:

Department-wise salary average
Subject-wise marks average

Concepts:

groupby()
mean()
sum()
count()

```python
import pandas as pd
data = {
    "Employee": ["Aarav", "Ananya", "Vivaan", "Diya", "Kabir", "Meera", "Rohan", "Isha", "Arjun", "Sanya"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance", "IT", "Finance", "HR", "IT"],
    "Salary": [85000, 60000, 90000, 75000, 65000, 80000, 88000, 72000, 62000, 92000],
    "Subject": ["Python", "Excel", "SQL", "Excel", "Excel", "Python", "Python", "SQL", "SQL", "Python"],
    "Marks": [88, 75, 92, 80, 85, 90, 78, 83, 89, 95]
}

df = pd.DataFrame(data)
print("Original Dataset:\n", df)
```

```python
dept_salary_avg = df.groupby('Department')['Salary'].mean()
print(dept_salary_avg)
```

```python
subject_marks_avg = df.groupby('Subject')['Marks'].mean()
print(subject_marks_avg)
```

```python
total_payroll = df.groupby('Department')['Salary'].sum()
print(total_payroll)
```

```python
subject_count = df.groupby('Subject')['Employee'].count()
print(subject_count)
```

Merge and Join DataFrames

Perform:

Inner join
Left join
Merge datasets

Concepts:

pd.merge()


```python
import pandas as pd
students_df = pd.DataFrame({
    "StudentID": [101, 102, 103, 104],
    "Name": ["Aarav", "Ananya", "Vivaan", "Diya"]
})
marks_df = pd.DataFrame({
    "StudentID": [101, 102, 103, 105],
    "Subject": ["Python", "Excel", "SQL", "Python"],
    "Marks": [88, 95, 79, 92]
})
```

```python
inner_result = pd.merge(students_df, marks_df, on='StudentID', how='inner')
print(inner_result)
```

```python
left_result = pd.merge(students_df, marks_df, on='StudentID', how='left')
print(left_result)
```
