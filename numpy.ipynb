Array Reshaping

Convert:

1D → 2D
2D → 1D

Concepts:

reshape()
flatten()
ravel()

```python
import numpy as np
matrix = np.array([[1, 2, 3], 
                   [4, 5, 6]])

print("--- Using flatten() ---")
flat_arr = matrix.flatten()
flat_arr[0] = 99  
print("Flattened Array:", flat_arr)
print("Original Matrix (Unchanged):\n", matrix)

print("\n--- Using ravel() ---")
rav_arr = matrix.ravel()
rav_arr[0] = 99  
print("Raveled Array:", rav_arr)
print("Original Matrix (CHANGED!):\n", matrix)
```

Filtering Arrays

Extract:

Even numbers
Numbers > 50
Negative numbers

Concept:

Boolean masking

Example:

arr[arr > 50]

```python
import numpy as np
arr = np.array([12, -7, 55, 34, -2, 88, 15, -40, 61])

print("Original Array:", arr)
print("-" * 40)

mask_gt50 = arr > 50
filtered_gt50 = arr[mask_gt50]
print("1. Numbers > 50:")
print("   Mask:   ", mask_gt50)
print("   Result: ", filtered_gt50)

mask_even = (arr % 2 == 0)
filtered_even = arr[mask_even]
print("\n2. Even Numbers:")
print("   Result: ", filtered_even)

mask_negative = arr < 0
filtered_negative = arr[mask_negative]
print("\n3. Negative Numbers:")
print("   Result: ", filtered_negative)
```

Stack and Split Arrays

Perform:

Horizontal stack
Vertical stack
Array splitting

Concepts:

np.hstack()
np.vstack()
np.split()

```python
import numpy as np
a = np.array([[1,2],
              [3,4]])
b = np.array([[5,2],
              [6,8]])
h_stacked = np.hstack((a,b))
v_stacked =np.vstack((a,b))
print("Horizontal Stack", h_stacked)
print("Vertical Stack",v_stacked)
print("-"*40)
large_array = np.array([10,30,48,56,69,78])
sub_array = np.split(large_array,3)
print("Original array:",large_array)
print("\nAfter splitting into 3 sections:")
for i, sub in enumerate(sub_array):
    print(f"Sub-array {i}: {sub}")
```

Solve Linear Algebra Problems
Matrix multiplication
Determinant
Inverse matrix

Concepts:

np.dot()
np.linalg.det()
np.linalg.inv()

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])
result=np.dot(A,B)
print(result)
det=np.linalg.det(A)
print(det)
inverse=np.linalg.inv(B)
print(inverse)
```

Mini NumPy Data Analysis Project

Create random student marks dataset and perform:

Average marks
Highest marks
Lowest marks
Students scoring above 80
Subject-wise analysis

Concepts:

Real-world array operations
Aggregation
Filtering
Axis operations

```python
import numpy as np

np.random.seed(1)

marks = np.random.randint(40, 101, size=(5, 3))

print("Student Marks Dataset:")
print(marks)

# Average
print("\nAverage Marks:")
print(np.mean(marks))

# Highest
print("\nHighest Marks:")
print(np.max(marks))

# Lowest
print("\nLowest Marks:")
print(np.min(marks))

# Above 80
print("\nMarks Above 80:")
print(marks[marks > 80])

# Subject-wise Average
print("\nSubject-wise Average:")
print(np.mean(marks, axis=0))

# Student-wise Average
print("\nStudent-wise Average:")
print(np.mean(marks, axis=1))
```
